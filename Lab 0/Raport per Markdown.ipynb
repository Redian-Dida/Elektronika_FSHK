# 2. Baza teorike dhe disa nga perdorimet dhe elementet e markdown
Ky sistem na lejon procesin e perpunimit qe si tekstin, si formulat te na dalin sa me mire vizualisht.
Disa nga elementet qe mund te tregojme jane perdorimi per rradhitjen e titullit nga i pari tek i dyti ose i treti. Gjithashtu kemi mundesi te ndryshojme dhe llojin ose trashesine e shkrimit ku per trashesine perdorim **tekst i trashe** dhe, per shembull, per shkrim te pjerret perdorim *tekst i pjerret*. Ne rast se duam te rradhitim dicka, mund te perdorim Markdown ne kete forme ku secili numer rradhitet.

1. First item
2. Second item
3. Third item

Gjithashtu kemi dhe metoden pa rradhitje ose ndryshe rastesore qe kryhet ne kete forme.

- First item
- Second item
- Third item

Keto ishin vetem disa nga perdorimet dhe mundesite qe na ofron Markdown.


# 3. Disa nga veprimet qe kryem me markdown
Perpos provave qe kryem duke ndryshuar tekstet, gjithashtu ne krijuam tabel dhe shkruajtem disa formula ku mund te permendim formulat e Maksuellit.
Shembull:

| Nr. | Emri | Mbiemri | Vendlindja |
| :---: | :--- | :--- | :--- |
| 1 | Arben | Hoxha | Tirane |
| 2 | Elona | Berisha | Durres |
| 3 | Genc | Krasniqi | Prishtine |

Gjithashtu nje shembull tjeter do ishin 4 formulat e Maksuellit qe trajtuam:

- **Ligji i Gausit per fushen elektrike:**
  $$\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}$$

- **Ligji i Gausit per fushen magnetike:**
  $$\nabla \cdot \mathbf{B} = 0$$

- **Ligji i Faradeit i induksionit:**
  $$\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}$$

- **Ligji i Amper-Maksuellit:**
  $$\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}$$

Keto jane disa nga menyrat qe trajtuam per perdorimin e Markdown.
