# 1. Qellimi i Eksperimentit
Ne kete eksperiment kam analizuar nje qark elektronik te perbere nga gjeneratori i sinjaleve, oshiloskopi dixhital dhe kater kondensatore me vlera te ndryshme. Qellimi kryesor ka qene vrojtimi dhe matja e parametrave te sinjaleve valore periodike, duke percaktuar tensionin, kohen dhe periudhen per llogaritjen e frekuences se sinjalit ne oshiloskop ($1250 \text{ Hz}$). Gjithashtu, kam studiuar sjelljen e qarqeve me kondensatore duke regjistruar periudhat e tyre ne shkallen e mikrosekondave ($10^{-6} \text{ s}$) dhe frekuencat perkatese, duke verifikuar marredheniet teorike mes tyre ne tryezen e punes.

## 2. Baza Teorike

### 2.1 Modeli i Qarkut
* **Skema e qarkut:** Per realizimin e dy eksperimenteve, lidha gjeneratorin e sinjaleve dhe oshiloskopin dixhital me qarkun perkates.
* **Sinjali hyres ($V_{in}$):** Perdora nje sinjal sinusoidal te dhene nga gjeneratori i funksioneve per te ekzaminuar lekundjet valore dhe sjelljen e qarkut me kondensatore.
* **Sinjali dales ($V_{out}$):** Regjistrova sinjalin ne skajet e qarkut permes oshiloskopit gjate matjeve te mia.
* **Vlerat e komponenteve:** 
  * Per Eksperimentin 1: Kryeva matjet e lekundjeve valore ku ndarjet ishin $2 \times 2$ ($V = 4$) dhe koha $0.2 \text{ ms}$ (periudha totale $T = 0.8 \text{ ms}$).
  * Per Eksperimentin 2: Bera matjet me kater kondensatore te ndryshem me periudhe $T$ ne shkallen e mikrosekondave ($10^{-6} \text{ s}$), konkretisht $3 \times 10^{-6} \text{ s}$, $6 \times 10^{-6} \text{ s}$, $2 \times 10^{-6} \text{ s}$ dhe $1 \times 10^{-6} \text{ s}$.
* **Supozimet:** I konsiderova komponentet ne kushte ideale dhe pranova se qarku funksionon ne regjim linear.

### 2.2 Ekuacionet Themelore
Per te analizuar te dhenat e mia, perdora lidhjen midis frekuences ($f$) dhe periudhes ($T$) per te dyja eksperimentet sipas formules:
$$f = \frac{1}{T}$$
Gjithashtu, per analizen e qarkut perdora raportin e transferit:
$$H(j\omega) = \frac{V_{out}}{V_{in}}$$
* **Vlera teorike ($f_{teori}$):** 
  * Per Eksperimentin 1: Llogarita frekuencen per periudhen $0.8 \text{ ms}$ e cila rezulton ne $1250 \text{ Hz}$.
  * Per Eksperimentin 2: Llogarita frekuencat per secilin nga kater kondensatoret bazuar ne vlerat e periudhave $T$ ($10^{-6} \text{ s}$), te cilat japin frekuencat perkatese $165.6 \text{ kHz}$, $337.4 \text{ kHz}$, $510 \text{ kHz}$ dhe $1 \text{ kHz}$.

# 4. Te Dhenat e Matura

# Pjesa 1: Matjet me Oshiloskop
Gjate vrojtimit te lekundjeve valore ne oshiloskop, kam regjistruar keto te dhena:
* Ndarjet e tensionit: $2 \times 2 \rightarrow V = 4$
* Koha per ndarje: $0.2 \text{ ms}$
* Periudha totale ($T$): $T = 4 \times 0.2 = 0.8 \text{ ms}$
* Frekuenca e llogaritur ($F$): 
  $$F = \frac{1}{0.8} \times 1000 = 1250 \text{ Hz}$$

# Pjesa 2: Matjet me 4 Kondensatore
Per kater kondensatoret e provuar ne qark, kam regjistruar periudhen ($T$) ne sekonda dhe frekuencat perkatese ($f$):

| Kondensatori 2 | $6 \times 10^{-6} \text{ s}$ | $337.4 \text{ kHz}$ |
