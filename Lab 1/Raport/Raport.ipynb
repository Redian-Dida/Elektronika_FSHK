**6. Analiza e Gabimeve**

Llogaritjen e gabimit relativ e kam kryer duke krahasuar vlerat e matura me ato nominale ose pritshmëritë teorike të qarkut:

$$\text{Gabimi} = \frac{\vert V_{exp} - V_{teori} \vert}{V_{teori}} \times 100\%$$

Duke u bazuar te matjet e mia (rezistencat $120\text{ k}\Omega$ dhe $1.77\text{ k}\Omega$, tensioni i burimit $11.69\text{ V}$, rryma në seri $0$ dhe në paralel $2.256\text{ A}$), llogaritjet numerike që kam bërë janë si më poshtë:

 1. Gabimi për tensionin në paralel:
  Për tensionin e burimit $V_{\text{teori}} = 11.69\text{ V}$ dhe tensionin e matur në paralel $V_{\text{exp}} = 11.36\text{ V}$:
  $$\text{Gabimi}_{\text{paralel}} = \frac{\vert 11.36 - 11.69 \vert}{11.69} \times 100\% = \frac{0.33}{11.69} \times 100\% \approx 2.82\%$$

 2. Gabimi për tensionin në seri:
  Për tensionin e matur në seri $V_{\text{exp}} = 3.93\text{ V}$ krahasuar me vlerën e llogaritur teorikisht $V_{\text{teori}} \approx 11.52\text{ V}$:
  $$\text{Gabimi}_{\text{seri}} = \frac{\vert 3.93 - 11.52 \vert}{11.52} \times 100\% = \frac{7.59}{11.52} \times 100\% \approx 65.88\%$$

Tolerancat e komponentëve: Rezistencat e përdorura mund të kenë devijime nga vlera e tyre nominale, gjë që ndikon drejtpërdrejt në shpërndarjen e rrymës dhe tensionit.

Pasiguritë e matjes: Leximet e multimetrit gjatë eksperimentit mund të shfaqin luhatje të vogla për shkak të ndjeshmërisë së instrumentit.
 
Kufizimet e instrumenteve: Gabimet e brendshme të multimetrit dhe të furnizuesit laboratorik gjatë matjeve të qarqeve në seri dhe paralel.
 
Efektet parazitare: Rezistenca e kontakteve në breadboard dhe lidhjet e telave, të cilat ishin veçanërisht të ndjeshme në qarkun në seri ku rryma doli konstante në vlerën $0$.
