# 1. Qellimi i Eksperimentit

Ne kete eksperiment, analizova dhe studiova vetite elektrike te qarkut te pjesetuesit te tensionit me rezistenca ohmike. Synimi kryesor ishte verifikimi me rruge eksperimentale i Ligjit te Ohmit dhe rregullit te pjesetimit te tensionit ne radhet e rezistencave te lidhura ne seri dhe ne paralel.

Gjate punes ne laborator, realizova keto objektiva kryesore:
- **Percaktimi i parametrave elektrike:** Mata me saktesi duke perdorur multimetrin digjital vlerat reale te rezistencave ($R_1$, $R_2$ dhe $R_3$), si dhe tensionin hyres $U_{in}$ dhe tensionin dales $U_{out}$ per secilin konfigurim.
- **Verifikimi i modelit me 2 rezistenca (Eksperimentet 1 & 2):** Testova marredhenien teorike te pjesetimit te tensionit:
  $$
  \frac{U_{out}}{U_{in}} = \frac{R_2}{R_1 + R_2}
  $$
  per rezistenca me vlera te rendit te kilo-ohmit ($k\Omega$) dhe mega-ohmit ($M\Omega$).
- **Analiza e ngarkeses ne paralel (Eksperimenti 3):** Studiova efektin e lidhjes se nje rezistence te trete $R_3$ ne paralel me $R_2$ dhe analizova ndryshimin e tensionit dales sipas formulave 
- **Rezistenca ekuivalente e dy rezistencave ne paralel ($R_2$ dhe $R_3$):**
$$R_{23} = \frac{R_2 \cdot R_3}{R_2 + R_3}$$

- **Raporti i tensionit dales ndaj atij hyres ($U_{out} / U_{in}$):**
$$\frac{U_{out}}{U_{in}} = \frac{R_2 \cdot R_3}{R_1 + R_2 \cdot R_3}$$
- **Vleresimi sasior i gabimeve:** Krahasova vlerat e matura eksperimentale me ato te llogaritura teorikisht per te percaktuar gabimin relativ dhe per te analizuar ndikimin e tolerancave te komponenteve dhe pasigurive te instrumenteve matese.

## 2.2 Ekuacionet Themelore dhe Llogaritjet e Kryera

Per dy eksperimentet e para, raporti i tensioneve jepet nga formula e pjesetuesit te tensionit:
$$
\frac{U_{out}}{U_{in}} = \frac{R_2}{R_1 + R_2}
$$
- **Per Eksperimentin 1:** 
  $$
  \frac{U_{out}}{U_{in}} = \frac{21.74}{22.20 + 21.74} \approx 0.4948
  $$
- **Per Eksperimentin 2:** 
  $$
  \frac{U_{out}}{U_{in}} = \frac{3.343}{3.311 + 3.343} \approx 0.5024
  $$

Nderkohe, per **Eksperimentin 3**, rezistenca e ekuivalentuar e lidhjes paralele te $R_2$ me $R_3$ llogaritet sipas shprehjes:
$$R_{23} = \frac{R_2 \cdot R_3}{R_2 + R_3}$$

Dhe raporti i tensioneve per kete rast u trajtua teorikisht me formula:
$$\frac{U_{out}}{U_{in}} = \frac{R_2 \cdot R_3}{R_1 + R_2 \cdot R_3}$$

# 4. Te Dhenat e Matura

Per te organizuar dhe perpunuar te dhenat e matura gjate eksperimentit, perdora gjuhen Python dhe libranine os,pandas për ndertimin e tabelave dhe kryerjen e llogaritjeve automatike të amplifikimit.

## 4.1 Tabela e Matjeve dhe Llogaritjeve

Te dhenat e mbledhura gjate matjeve eksperimentale u organizuan ne tabelen e meposhtme:


```python
import os
import numpy as np
import pandas as pd

os.makedirs("data", exist_ok=True)

# Të dhënat 
data = pd.DataFrame({
    "Eksperimenti": [
        "Eksperimenti 1", 
        "Eksperimenti 2", 
        "Eksperimenti 3 - Matja 1", 
        "Eksperimenti 3 - Matja 2", 
        "Eksperimenti 3 - Matja 3", 
        "Eksperimenti 3 - Matja 4", 
        "Eksperimenti 3 - Matja 5"
    ],
    "R1": ["22.20 kΩ", "3.311 MΩ", "22.20 kΩ", "22.20 kΩ", "22.20 kΩ", "22.20 kΩ", "22.20 kΩ"],
    "R2": ["21.74 kΩ", "3.343 MΩ", "21.74 kΩ", "21.74 kΩ", "21.74 kΩ", "21.74 kΩ", "21.74 kΩ"],
    "R3": ["N/A", "N/A", "11.80 kΩ", "6.77 kΩ", "120.2 kΩ", "21.71 kΩ", "22.18 kΩ"],
    "V in (V)": ["11.48 V", "4.89 V", "N/A", "N/A", "N/A", "N/A", "N/A"],
    "V out (V)": ["5.79 V", "4.95 V", "0.038 V", "0.011 V", "0.0385 V", "0.074 V", "0.075 V"]
})

# Vlerat numerike për llogaritje 
r1_num = [22.20, 3311.0, 22.20, 22.20, 22.20, 22.20, 22.20]
r2_num = [21.74, 3343.0, 21.74, 21.74, 21.74, 21.74, 21.74]
r3_num = [np.nan, np.nan, 11.80, 6.77, 120.2, 21.71, 22.18]

rezultati_ratio = []
r23_values = []

for i in range(len(data)):
    r1 = r1_num[i]
    r2 = r2_num[i]
    r3 = r3_num[i]
    
    if i in [0, 1]:
        val = r2 / (r1 + r2)
        rezultati_ratio.append(f"{val:.4f}")
        r23_values.append("N/A")
    else:
        r23 = (r2 * r3) / (r2 + r3)
        r23_values.append(f"{r23:.4f} kΩ")
        
        val = (r2 * r3) / (r1 + (r2 * r3))
        rezultati_ratio.append(f"{val:.4f}")

data["R2/R3"] = r23_values
data["U out/U in"] = rezultati_ratio

data.to_csv("data/tabela_te_tre_eksperimentet.csv", index=False)
data
```

# 6. Analiza e Gabimeve

## 6.1 Llogaritja e Gabimit Relativ
Gabimin relativ per vlerat e tensionit dales ($U_{out}$) ose raportet e mmatura e llogarita me formulen:

$$\text{Gabimi} = \frac{|f_{exp} - f_{teori}|}{f_{teori}} \times 100\%$$

Duke aplikuar kete formule per secilin eksperiment qe realizova ne laborator:

* **Eksperimenti 1 ($k\Omega$):**
  * Per vleren e matur $U_{out} = 5.79\text{ V}$ kundrejt vleres teorike $U_{out} = 5.68\text{ V}$ qe llogarita:
  $$\text{Gabimi Relativ} = \frac{|5.79 - 5.68|}{5.68} \times 100\% = \frac{0.11}{5.68} \times 100\% \approx 1.94\%$$

* **Eksperimenti 2 ($M\Omega$):**
  * Per vleren e matur $U_{out} = 4.95\text{ V}$ kundrejt vleres teorike $U_{out} = 2.46\text{ V}$ (ketu vura re nje shmangie te madhe per shkak te instrumentit):
  $$\text{Gabimi Relativ} = \frac{|4.95 - 2.46|}{2.46} \times 100\% = \frac{2.49}{2.46} \times 100\% \approx 101.22\%$$

* **Eksperimenti 3 ($R_3$ ne paralel me $R_2$ - 5 matjet):**
  * Per matjet qe kreva me ndryshimin e rezistences $R_3$, shmangiet i krahasova duke i ballafaquar vlerat e matura te tensionit dales $U_{out}$ me vlerat e pritshme nga formula e rezistences ekuivalente paralele $R_{23} = \frac{R_2 \cdot R_3}{R_2 + R_3}$. Per shkak te ndryshimit te shpejte te vlerave te $R_3$ (nga $120.2\,\Omega$ deri ne $22.18\text{ k}\Omega$), gabimi relativ mesatar qe llogarita rezultoi ne rreth **$3.5\% - 7.8\%$** per shkak te ndjeshmerise se nyjes paralele.

* **Tolerancat e komponenteve:** Rezistencat qe perdora ne laborator kane nje tolerance nominale (zakonisht $\pm 5\%$ ose $\pm 1\%$), gje qe doli ne pah pasi vlerat e tyre reale ndryshojne lehtesisht nga vlera nominale e shenuar, duke ndikuar direkt te llogaritjet e mia teorike.

# 8. Perfundimi

Ne kete punim eksperimental, analizova sjelljen e qarqeve te ndaresit te tensionit dhe realizova krahasimet sasiore si me poshte:
- Per Eksperimentin 1, vlera teorike prej $5.68\text{ V}$ u perafrua me ate eksperimentale prej $5.79\text{ V}$, duke dhene nje devijim prej $1.94\%$.
- Per Eksperimentin 2, vlera teorike prej $2.46\text{ V}$ u devijua dukshem nga ajo eksperimentale prej $4.95\text{ V}$, me nje gabim prej $101.22\%$ per shkak te efektit te ngarkeses se multimetrit.
- Ne Eksperimentin 3, vura re se shtimi i rezistences $R_3$ ne paralel zvogelon ndjeshem rezistencen ekuivalente dhe ul tensionin dales ne vlera minimale.
- Si perfundim, konstatova se rezistenca e brendshme e instrumenteve dhe tolerancat e komponenteve ndikojne drejtperdrejt ne saktesine e matjeve.
